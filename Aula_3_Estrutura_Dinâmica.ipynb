**Importante:** *Python não possui ponteiros explícitos como C/C++. Nesta aula, o conceito de ponteiro é usado para estabelecer a base conceitual para referências e estruturas dinâmicas.*

```python
# Exercício 1 — Referências

a = [10, 20, 30]
b = a

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
```

Responda:

▶ **Qual será a saída?**

a: [10, 20, 30, 40]
b: [10, 20, 30, 40]
id(a): 140358913214848 (Este número varia a cada execução)
id(b): 140358913214848 (Será exatamente o mesmo número que o de "a")

▶ **a e b representam o mesmo objeto?**

SIM

▶ **Por quê?**

Em Python, variáveis são como *etiquetas* que apontam para um local na memória. Quando você faz b = a, você não está copiando os dados da lista, mas sim criando uma nova etiqueta (b) que aponta para o mesmo objeto na memória que a. Por isso, ao alterar a lista através de b (usando o append), a lista a também reflete a mudança, pois ambas são a mesma coisa. O fato de id(a) e id(b) serem iguais comprova isso.

▶ **O que aconteceria com b = a.copy()?**

Neste caso, a saída seria diferente. O método .copy() cria um novo objeto na memória com os mesmos valores da lista original (uma cópia rasa). Se você fizesse isso:

* b receberia o valor 40, ficando [10, 20, 30, 40].

* a permaneceria inalterado: [10, 20, 30].

id(a) e id(b) seriam números diferentes, comprovando que são dois objetos distintos na memória.

```python
# Exercício 2 — Construindo uma cadeia

# Crie:

# n1 = Node("A")
# n2 = Node("B")
# n3 = Node("C")

# Conecte:

# A → B → C → None
# Depois percorra a estrutura e produza:

# A
# B
# C

# 1. Definindo a classe Node (necessário para o código funcionar)
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None

# 2. Criando os nós
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

# 3. Conectando A -> B -> C -> None
n1.proximo = n2
n2.proximo = n3
# Nota: n3.proximo já é None por padrão devido ao __init__

# 4. Percorrendo a estrutura
atual = n1
while atual is not None:
    print(atual.valor)
    atual = atual.proximo  # Avança para o próximo nó
```

**Perguntas:**

▶ **Qual é o problema?:**
O programa contém um loop infinito (laço infinito).

▶ **Por que o programa não termina?**
Dentro do bloco while, a variável atual nunca sofre alteração. Como ela é inicializada apontando para n1 e a condição é atual is not None (o que é verdadeiro, pois n1 não é nulo), o programa fica imprimindo o valor 10 repetidamente, travando a execução.

▶ **Qual linha deve ser acrescentada?**
Você precisa fazer a variável avançar para o próximo nó da cadeia a cada repetição. A seguinte linha deve ser acrescentada dentro do while, logo abaixo do print:

* atual = atual.proximo
