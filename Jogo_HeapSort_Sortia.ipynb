```python
#!/usr/bin/env python3
"""Jogo didático de Heapsort para execução no terminal.

O estudante constrói uma max-heap e depois executa o Heapsort escolhendo
manualmente as trocas. O programa valida cada decisão e explica os índices.
Não usa bibliotecas externas.
"""

from __future__ import annotations

import random
from dataclasses import dataclass


@dataclass
class Placar:
    """Armazena as informações do desempenho do jogador durante a atividade."""
    acertos: int = 0
    erros: int = 0
    dicas: int = 0

    @property
    def pontos(self) -> int:
        # Cada acerto vale 10 pontos. Erros (2 pts) e dicas (1 pt) reduzem a pontuação.
        # A função max() impede que o resultado final seja negativo.
        return max(0, (self.acertos * 10) - (self.erros * 2) - self.dicas)


def filhos(indice: int, tamanho: int) -> list[int]:
    """Retorna os índices dos filhos existentes de um nó."""
    # Em uma heap representada por vetor, os filhos de um nó de índice i são:
    # filho esquerdo = 2*i + 1 e filho direito = 2*i + 2.
    candidatos = (2 * indice + 1, 2 * indice + 2)

    # Retorna somente os filhos que realmente existem dentro da heap ativa.
    return [i for i in candidatos if i < tamanho]


def maior_filho(vetor: list[int], indice: int, tamanho: int) -> int | None:
    """Retorna o índice do maior filho, priorizando o esquerdo em empate."""
    indices = filhos(indice, tamanho)

    # Se houver filhos, procura aquele cujo valor no vetor é o maior.
    # Caso o nó não tenha filhos, retorna None.
    return max(indices, key=lambda i: vetor[i]) if indices else None


def tem_propriedade_heap(vetor: list[int], indice: int, tamanho: int) -> bool:
    """Verifica a propriedade de max-heap em um nó."""
    # Em uma max-heap, o pai deve possuir valor maior ou igual ao de todos os seus filhos.
    return all(vetor[indice] >= vetor[f] for f in filhos(indice, tamanho))


def eh_max_heap(vetor: list[int], tamanho: int) -> bool:
    """Verifica se o prefixo vetor[:tamanho] é uma max-heap."""
    # Apenas os nós da primeira metade precisam ser verificados, pois
    # os elementos restantes são folhas e não possuem filhos.
    return all(tem_propriedade_heap(vetor, i, tamanho) for i in range(tamanho // 2))


def exibir_arvore(vetor: list[int], tamanho_heap: int) -> None:
    """Mostra a heap por níveis e separa a região já ordenada."""
    # Exibe cada elemento no formato [índice:valor].
    print("\nVetor:", " ".join(f"[{i}:{v}]" for i, v in enumerate(vetor)))

    # Durante o Heapsort, o vetor é dividido em duas regiões:
    # heap ativa e elementos que já estão em suas posições definitivas.
    print("Heap ativa:", vetor[:tamanho_heap], "| Ordenados:", vetor[tamanho_heap:])

    if tamanho_heap == 0:
        # Se não existir mais heap ativa, não há árvore para exibir.
        return

    # A heap é armazenada em vetor, mas pode ser visualizada como árvore.
    print("Árvore lógica (índice:valor):")
    inicio, largura = 0, 1

    while inicio < tamanho_heap:
        fim = min(inicio + largura, tamanho_heap)
        print("  " + "   ".join(f"{i}:{vetor[i]}" for i in range(inicio, fim)))
        # Avança para o próximo nível da árvore (o dobro de nós do nível anterior).
        inicio, largura = fim, largura * 2


def ler_escolha(prompt: str, opcoes: set[str]) -> str:
    """Lê a entrada do usuário até que uma opção válida seja fornecida."""
    while True:
        resposta = input(prompt).strip().lower()
        # A função só termina quando o usuário informa uma opção permitida.
        if resposta in opcoes:
            return resposta
        print("Opção inválida. Escolha:", ", ".join(sorted(opcoes)))


def perguntar_troca(
    vetor: list[int], indice_pai: int, indice_filho: int, placar: Placar
) -> None:
    """Solicita a troca correta entre pai e maior filho."""
    while True:
        print(f"O nó {indice_pai}:{vetor[indice_pai]} viola a propriedade de max-heap.")
        resposta = input(
            "Digite os índices que devem ser trocados (ex.: 1 4) ou 'dica': "
        ).strip().lower()

        # O jogador pode solicitar ajuda sem realizar uma tentativa de troca.
        if resposta == "dica":
            placar.dicas += 1
            print("Dica: compare os dois filhos e troque o pai com o filho de maior valor.")
            continue

        try:
            # Converte os dois índices digitados para números inteiros.
            a, b = map(int, resposta.split())
        except ValueError:
            print("Informe exatamente dois índices inteiros.")
            continue

        # O uso de conjuntos permite aceitar os índices em qualquer ordem.
        if {a, b} == {indice_pai, indice_filho}:
            # Troca os valores das duas posições diretamente no vetor.
            vetor[indice_pai], vetor[indice_filho] = vetor[indice_filho], vetor[indice_pai]
            placar.acertos += 1
            print("✓ Troca correta!")
            return

        placar.erros += 1
        print("✗ Essa troca não restaura a propriedade de heap neste nó. Tente novamente.")


def descer_manual(vetor: list[int], raiz: int, tamanho: int, placar: Placar) -> None:
    """Restaura a heap, exigindo que o estudante escolha cada troca."""
    # Começa pelo nó indicado como raiz da subárvore analisada.
    pai = raiz

    while True:
        # Localiza o filho de maior valor.
        filho = maior_filho(vetor, pai, tamanho)

        # A descida termina se o nó não possuir filhos ou se o pai já for maior ou igual ao maior filho.
        if filho is None or vetor[pai] >= vetor[filho]:
            print(f"O nó {pai}:{vetor[pai]} já possui a propriedade de heap.")
            return

        # Se o pai for menor que o maior filho, a propriedade da max-heap foi violada.
        exibir_arvore(vetor, tamanho)
        perguntar_troca(vetor, pai, filho, placar)

        # Depois da troca, continua verificando a posição para onde o antigo valor do pai desceu.
        pai = filho


def construir_heap(vetor: list[int], placar: Placar) -> None:
    """Fase 1: Transforma o vetor em uma max-heap."""
    print("\n=== FASE 1 — CONSTRUIR A MAX-HEAP ===")
    print("Percorreremos os pais de baixo para cima.")

    # O último nó que pode possuir filhos está em len(vetor)//2 - 1.
    # A construção começa nele e segue até a raiz, índice 0.
    for pai in range(len(vetor) // 2 - 1, -1, -1):
        print(f"\nAnalisando o nó de índice {pai}...")
        descer_manual(vetor, pai, len(vetor), placar)

    # Confirma internamente que a construção foi realizada corretamente.
    assert eh_max_heap(vetor, len(vetor))
    exibir_arvore(vetor, len(vetor))
    print("✓ Max-heap construída: o maior valor está na raiz.")


def ordenar_manual(vetor: list[int], placar: Placar) -> None:
    """Fase 2: Utiliza a max-heap construída para ordenar os elementos."""
    print("\n=== FASE 2 — ORDENAR ===")

    # 'fim' representa a última posição pertencente à heap ativa.
    # A cada repetição, essa região diminui uma posição.
    for fim in range(len(vetor) - 1, 0, -1):
        exibir_arvore(vetor, fim + 1)
        print("A raiz contém o maior valor da heap ativa.")

        while True:
            resposta = input(
                f"Quais índices devem ser trocados para fixar o máximo na posição {fim}? "
            ).strip().lower()

            if resposta == "dica":
                placar.dicas += 1
                print(f"Dica: troque a raiz (índice 0) com o último índice ativo ({fim}).")
                continue

            try:
                a, b = map(int, resposta.split())
            except ValueError:
                print("Digite dois índices ou 'dica'.")
                continue

            # O maior elemento (raiz) deve ser trocado com o último elemento da região ativa.
            if {a, b} == {0, fim}:
                vetor[0], vetor[fim] = vetor[fim], vetor[0]
                placar.acertos += 1
                print("✓ Máximo colocado em sua posição definitiva!")
                break

            placar.erros += 1
            print("✗ A troca deve retirar a raiz e reduzir a heap ativa.")

        # Após retirar o maior elemento, a raiz pode deixar de respeitar a propriedade
        # da heap. Por isso, ela é reorganizada. 'fim' já está ordenado e não participa.
        descer_manual(vetor, 0, fim, placar)


def obter_vetor() -> list[int]:
    """Coleta os dados do vetor a ser ordenado, seja manual ou aleatório."""
    print("\n1 — Usar números aleatórios")
    print("2 — Digitar os números")
    escolha = ler_escolha("Escolha: ", {"1", "2"})

    if escolha == "1":
        # Gera sete valores diferentes entre 1 e 99.
        return random.sample(range(1, 100), 7)

    while True:
        try:
            entrada = input("Digite de 4 a 12 inteiros separados por espaço: ")
            # split() separa a entrada e int() converte cada parte para inteiro.
            valores = [int(x) for x in entrada.split()]

            # O jogo aceita vetores contendo entre 4 e 12 elementos.
            if 4 <= len(valores) <= 12:
                return valores
        except ValueError:
            # O erro ocorre se algum dos valores não puder ser convertido para int.
            pass
        print("Entrada inválida. Use somente inteiros e informe entre 4 e 12 valores.")


def main() -> None:
    print("=" * 58)
    print("JOGO DE HEAPSORT — APRENDER FAZENDO")
    print("=" * 58)

    # Apresenta ao jogador as regras fundamentais utilizadas para organizar a heap.
    print("Em um vetor com índice i: esquerdo = 2i+1; direito = 2i+2.")
    print("Regra da max-heap: todo pai deve ser maior ou igual aos filhos.")

    vetor = obter_vetor()
    original = vetor.copy()        # Necessário porque o Heapsort modifica o vetor.
    placar = Placar()              # Cria o placar (zero acertos, erros e dicas).

    construir_heap(vetor, placar)  # Fase 1: transforma o vetor em uma max-heap.
    ordenar_manual(vetor, placar)  # Fase 2: utiliza a max-heap para ordenar os elementos.

    exibir_arvore(vetor, 0)        # tamanho_heap = 0 indica que todos estão ordenados.

    # Verificação interna: compara o jogo com a ordenação nativa do Python.
    assert vetor == sorted(original)

    print("\nParabéns! Ordenação concluída.")
    print("Original:", original)
    print("Ordenado:", vetor)

    # Apresenta a pontuação e o desempenho final do estudante.
    print(
        f"Placar: {placar.pontos} pontos "
        f"({placar.acertos} acertos, {placar.erros} erros, {placar.dicas} dicas)"
    )


# Garante que main() execute apenas quando o arquivo for iniciado diretamente.
if __name__ == "__main__":
    main()
```
