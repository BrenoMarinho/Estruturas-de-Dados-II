**FIFO (First In, First Out)** — O Primeiro a Entrar é o Primeiro a SairEstrutura associada: Fila (Queue).  

Funcionamento: O primeiro elemento adicionado é obrigatoriamente o primeiro a ser removido.  

Inserção e Remoção: Novos elementos entram no final da fila, enquanto as remoções acontecem no início.  

Exemplo prático: Uma fila de banco ou de impressão — o primeiro documento enviado para impressão é o primeiro a ser processado.

**LIFO (Last In, First Out**) — O Último a Entrar é o Primeiro a SairEstrutura associada: Pilha (Stack).  

Funcionamento: O elemento adicionado mais recentemente é o primeiro a ser removido.  

Inserção e Remoção: Tanto a inserção (push) quanto a remoção (pop) ocorrem exclusivamente na mesma extremidade: o topo.  

Exemplo prático: Uma pilha de pratos ou o histórico de navegação ("Voltar") de um navegador — a última página acessada é a primeira a ser fechada ao voltar.

Exercício 1 | Simulador "Desfazer"  

Objetivo: Implemente uma Pilha para gerenciar ações em um editor de texto.

Requisitos: Armazene cada ação (digitar, apagar, substituir). Ao acionar a função 'desfazer', o programa deve remover e reverter estritamente o último comando inserido

```python
class EditorTexto:
    def __init__(self):
        # Inicializa uma lista vazia que funcionará como nossa estrutura de Pilha.
        # Em Python, listas são excelentes para simular pilhas devido aos métodos append() e pop().
        self.historico_acoes = []  # Pilha (LIFO - Last In, First Out)

    def executar_acao(self, acao: str):
        # O método append() adiciona a nova ação no "topo" da pilha (no final da lista).
        self.historico_acoes.append(acao)
        print(f"Ação realizada: '{acao}'")

    def desfazer(self):
        # Primeiro, verifica se a lista está vazia para evitar um erro (IndexError)
        # ao tentar fazer pop() em uma lista sem elementos.
        if not self.historico_acoes:
            print("Nenhuma ação para desfazer.")
            return None

        # O método pop() remove e retorna o ÚLTIMO elemento adicionado à lista.
        # É isso que garante o comportamento LIFO (o último a entrar é o primeiro a sair).
        acao_desfeita = self.historico_acoes.pop()
        print(f"Desfeito: '{acao_desfeita}'")
        return acao_desfeita

# ==========================================
# Teste de execução do código
# ==========================================
editor = EditorTexto()

# Empilhando as ações. Cada nova ação vai para o final da lista.
editor.executar_acao("Digitar 'Olá'")    # A pilha agora é: ["Digitar 'Olá'"]
editor.executar_acao("Digitar ' Mundo'") # A pilha agora é: ["Digitar 'Olá'", "Digitar ' Mundo'"]
editor.executar_acao("Apagar ' Mundo'")  # A pilha agora é: ["Digitar 'Olá'", "Digitar ' Mundo'", "Apagar ' Mundo'"]

print("-" * 30) # Apenas um separador visual para o terminal

# Desempilhando as ações. O pop() vai sempre retirar a última que foi inserida.
editor.desfazer()  # Retira "Apagar ' Mundo'". Sobram as duas primeiras.
editor.desfazer()  # Retira "Digitar ' Mundo'". Sobra apenas a primeira.
```

Exercício 2 | Sistema de Impressão  

Objetivo: Implemente uma Fila para gerenciar um spooler de impressora.

Requisitos: Cada novo documento é enfileirado. O sistema deve garantir que o documento impresso seja sempre o mais antigo aguardando na fila.

```python
# Importamos 'deque' (Double-Ended Queue) do módulo collections.
# Por baixo dos panos (no CPython), o deque é implementado como uma lista
# duplamente encadeada. Isso o torna incrivelmente rápido para manipular as pontas.
from collections import deque

class Impressora:
    def __init__(self):
        # Inicializa a fila.
        # O comportamento garantido aqui é FIFO (Primeiro a entrar, primeiro a sair).
        self.fila_impressao = deque()

    def adicionar_documento(self, documento: str):
        # Enfileira o documento inserindo-o no final (à direita) do deque.
        # Operação de complexidade O(1) - tempo constante, independente do tamanho da fila.
        self.fila_impressao.append(documento)
        print(f"Documento '{documento}' adicionado à fila de impressão.")

    def imprimir_documento(self):
        # Validação de segurança para evitar erro ao remover de uma estrutura vazia.
        if not self.fila_impressao:
            print("Nenhum documento na fila para imprimir.")
            return None

        # O método popleft() remove e retorna o elemento no início da fila (à esquerda).
        # É aqui que o 'deque' brilha: se usássemos uma lista normal com pop(0), a
        # complexidade seria O(n), pois o Python precisaria realocar todos os itens
        # restantes na memória. Com o deque, o popleft() é O(1), extremamente eficiente.
        documento_impresso = self.fila_impressao.popleft()
        print(f"Imprimindo: '{documento_impresso}'.")
        return documento_impresso

# ==========================================
# Teste de execução do código
# ==========================================
impressora = Impressora()

# Adiciona documentos na ordem: 1º Relatório, 2º Ata, 3º Fatura
impressora.adicionar_documento("Relatório Mensal")
impressora.adicionar_documento("Ata da Reunião")
impressora.adicionar_documento("Fatura Cliente X")

print("-" * 30)

# Imprime respeitando o FIFO: O mais antigo sai primeiro.
impressora.imprimir_documento() # Sai: 'Relatório Mensal'
impressora.imprimir_documento() # Sai: 'Ata da Reunião'

print("-" * 30)

# Simula um cenário dinâmico onde novos itens chegam durante o processamento.
impressora.adicionar_documento("E-mail Importante")

print("-" * 30)

# A impressão continua de onde parou (Fatura era o próximo da fila).
impressora.imprimir_documento() # Sai: 'Fatura Cliente X'
impressora.imprimir_documento() # Sai: 'E-mail Importante'

# Fila totalmente processada e vazia, aciona a validação.
impressora.imprimir_documento() # Indica que não há documentos
```

▶ **Desafio Master | Triagem Hospitalar**

Cenário: Criar um programa de gerenciamento para uma fila de atendimento médico.  

▪ Regras de Negócio:

Fluxo Padrão: Pacientes normais entram no final da fila.  

Fluxo de Exceção: Pacientes prioritários (idosos, emergências) devem furar a fila e entrar diretamente no início do atendimento.  

*Dica de Arquitetura: A estrutura collections.deque é uma fila de duas pontas. Explore os métodos de inserção à esquerda para resolver a prioridade sem reescrever a fila inteira!*

```python
class Pilha:
    def __init__(self, capacidade):
        # O __init__ é o construtor da classe.
        # Ele define o 'estado' inicial do nosso objeto.
        self.capacidade = capacidade

        # O atributo self.itens fica "escondido" dentro da classe.
        # A ideia do encapsulamento é que quem usa a Pilha não deve manipular
        # essa lista diretamente, mas sim usar os métodos push() e pop().
        self.itens = []

    def esta_cheia(self):
        # Método auxiliar (helper).
        # Em vez de espalhar "len(self.itens) == self.capacidade" pelo código,
        # centralizamos a regra de negócio aqui. Fica muito mais legível.
        return len(self.itens) == self.capacidade

    def esta_vazia(self):
        # Retorna um booleano (True ou False).
        return len(self.itens) == 0

    def push(self, valor):
        # A operação de inserção atua como um "porteiro", validando a regra
        # de negócio (limite de capacidade) antes de alterar o estado.
        if self.esta_cheia():
            print("Stack Overflow! Pilha cheia.")
            # O 'return' vazio funciona como um break, impedindo que
            # a execução chegue no append() abaixo.
            return

        self.itens.append(valor)
        print(f"Push: {valor} inserido no topo.")

    def pop(self):
        # Novamente, a validação de estado acontece antes da ação.
        if self.esta_vazia():
            print("Stack Underflow! Pilha vazia.")
            # RETORNO SENTINELA: Retorna -1 para sinalizar à parte do programa
            # que chamou essa função que ocorreu um erro (já que não há valor a retornar).
            return -1

        valor_removido = self.itens.pop()
        print(f"Pop: {valor_removido} removido do topo.")
        return valor_removido

# ==========================================
# Teste de execução do código
# ==========================================
p = Pilha(capacidade=5)

# O estado da pilha muda através de métodos controlados (Encapsulamento).
p.push(10)
p.push(20)
p.push(30)

print("-" * 30)

p.pop()
p.pop()

print("-" * 30)

p.push(40)

print("-" * 30)

p.pop()
p.pop()
p.pop() # Dispara o alerta de Stack Underflow e retorna -1.
```

 01 | A Pilha Matemática  Missão: Crie uma pilha (LIFO) que armazene números inteiros. Escreva um algoritmo que esvazie a pilha iterativamente, calculando e retornando a soma total dos elementos removidos.

```python
class PilhaMatematica:
    def __init__(self):
        # A pilha continua sendo iniciada como uma lista vazia.
        self.elementos = []

    def push(self, item):
        # Adiciona o elemento no topo.
        self.elementos.append(item)
        print(f"Adicionado {item} à pilha.")

    def pop(self):
        # Melhoria em relação ao código anterior:
        # Retornar 'None' em vez de '-1' é uma prática muito mais segura em Python
        # quando não queremos usar Exceptions, pois evita somar números negativos por engano.
        if not self.elementos:
            return None # Sinaliza que a pilha está vazia

        item_removido = self.elementos.pop()
        return item_removido

    def esta_vazia(self):
        # Retorna True se o tamanho for 0.
        return len(self.elementos) == 0

# ==========================================
# Função para processar (e consumir) a pilha
# ==========================================
def somar_elementos_pilha(pilha):
    soma_total = 0
    print("\nEsvaziando a pilha e somando os elementos...")

    # Este é o padrão clássico para esvaziar/processar pilhas e filas.
    # Enquanto a pilha não estiver vazia, o loop continua rodando.
    while not pilha.esta_vazia():
        # A cada iteração, retiramos o topo da pilha.
        # Isso significa que estamos "consumindo" a estrutura de dados.
        elemento = pilha.pop()

        # O 'is not None' é uma camada de segurança extra.
        # Como o loop while já verifica se a pilha tem itens, teoricamente
        # nunca receberíamos um None aqui, mas é uma boa prática de programação defensiva.
        if elemento is not None:
            soma_total += elemento
            print(f"Removido {elemento}. Soma atual: {soma_total}")

    # Quando o while termina, a pilha original está completamente vazia.
    return soma_total

# ==========================================
# Teste da Pilha Matemática
# ==========================================
minha_pilha = PilhaMatematica()

# Empilha os valores: [5, 10, 3, 7] (7 fica no topo)
minha_pilha.push(5)
minha_pilha.push(10)
minha_pilha.push(3)
minha_pilha.push(7)

# Ao chamar essa função, passamos a referência do objeto 'minha_pilha'.
# A ordem de soma será do topo para a base: 7 + 3 + 10 + 5 = 25.
soma_final = somar_elementos_pilha(minha_pilha)
print(f"\nSoma total de todos os elementos removidos: {soma_final}")

# ==========================================
# Teste com pilha vazia
# ==========================================
print("\n--- Teste com pilha inicialmente vazia ---")
pilha_vazia = PilhaMatematica()

# O while not pilha.esta_vazia() vai ser falso logo de cara.
# O loop não roda nenhuma vez, e a função retorna soma_total (que é 0).
soma_vazia = somar_elementos_pilha(pilha_vazia)
print(f"Soma total de elementos removidos de pilha vazia: {soma_vazia}")
```

02 | O Call Center  Missão: Crie uma fila (FIFO) que armazene nomes de clientes. Simule um ciclo contínuo de atendimento: adicionar novos clientes à espera e "chamar" o próximo cliente da fila para atendimento.

```python
# Continuamos usando o 'deque' por sua eficiência O(1) ao remover itens do início.
from collections import deque

class CallCenter:
    def __init__(self):
        # Inicializa a fila de espera dos clientes.
        self.fila_clientes = deque()

        # Este é um "contador de estado". Ele guarda o número do próximo cliente
        # anônimo. É a mesma lógica usada em bancos de dados (Auto Increment).
        self.proximo_id_cliente = 1

    def adicionar_cliente(self, nome=None):
        # Aqui temos o uso de um "Argumento Padrão" (nome=None).
        # Se quem chamar a função não passar um nome, o Python assume None.
        if nome is None:
            # Cria um nome genérico usando o contador atual.
            nome = f"Cliente_{self.proximo_id_cliente}"
            # Incrementa o contador para que o próximo anônimo tenha um ID diferente.
            self.proximo_id_cliente += 1

        # O cliente (com nome real ou gerado) entra no final da fila.
        self.fila_clientes.append(nome)
        print(f"Cliente '{nome}' entrou na fila. Clientes na fila: {len(self.fila_clientes)}")

    def chamar_proximo_cliente(self):
        # Regra de ouro: sempre verifique se a fila não está vazia antes de remover.
        if not self.fila_clientes:
            print("Nenhum cliente na fila de espera.")
            return None

        # popleft() garante a regra FIFO (Primeiro a entrar, primeiro a sair).
        # O cliente que esperou mais tempo (no início da fila) é atendido.
        cliente_chamado = self.fila_clientes.popleft()
        print(f"Chamando o próximo! Atendendo: '{cliente_chamado}'. Clientes restantes: {len(self.fila_clientes)}")
        return cliente_chamado

    def status_fila(self):
        # Converte o deque para uma lista normal apenas na hora de imprimir.
        # Isso facilita a visualização para fins de depuração (debug).
        print(f"Status da Fila: {list(self.fila_clientes)}")

# ==========================================
# Simulação do Call Center
# ==========================================
meu_call_center = CallCenter()

print("--- Chegada de Clientes ---")
# Clientes com nomes explícitos
meu_call_center.adicionar_cliente("Ana")
meu_call_center.adicionar_cliente("Bruno")

# Aqui a mágica acontece: não passamos nome. O sistema assume 'Cliente_1'.
meu_call_center.adicionar_cliente()

# Clientes com nomes voltam a entrar na fila normal.
meu_call_center.adicionar_cliente("Carla")

# Mostra: ['Ana', 'Bruno', 'Cliente_1', 'Carla']
meu_call_center.status_fila()

print("\n--- Ciclo de Atendimento ---")
# A fila é dinâmica! Atendemos a Ana...
meu_call_center.chamar_proximo_cliente()

# ...e um novo cliente chega enquanto o atendimento acontece.
meu_call_center.adicionar_cliente("Daniel")

# Atendemos o Bruno...
meu_call_center.chamar_proximo_cliente()

# Mostra quem sobrou: ['Cliente_1', 'Carla', 'Daniel']
meu_call_center.status_fila()

# Zera a fila chamando os três restantes
meu_call_center.chamar_proximo_cliente() # Atende Cliente_1
meu_call_center.chamar_proximo_cliente() # Atende Carla
meu_call_center.chamar_proximo_cliente() # Atende Daniel

# Tenta chamar alguém com a fila já vazia. O sistema intercepta o erro suavemente.
meu_call_center.chamar_proximo_cliente()
meu_call_center.status_fila()
```

**01 | Pilha via Array Estático**

 Missão: Implemente uma pilha em C alocando um array de tamanho fixo. Desenvolva as funções push() e pop(), criando uma variável de controle manual para rastrear o índice exato do topo (top).  


*Obs.: Devido ao ambiente ser o Google Colab, ele não permite o processamento do exercício em C, dessa forma, foi realizado uma adaptação para o processamento em Python.*


```python
class PilhaArrayEstatico:
    def __init__(self, capacidade):
        self.capacidade = capacidade

        # Em vez de uma lista vazia [], criamos uma lista já com o tamanho máximo definido.
        # [None, None, None] (se a capacidade for 3). Isso simula a alocação
        # estática de memória. A lista não vai crescer nem encolher.
        self.array = [None] * capacidade

        # O 'topo' funciona como um "ponteiro". Ele guarda o índice do último item inserido.
        # Começa em -1 porque os índices em Python começam em 0.
        # Portanto, -1 significa "antes do primeiro espaço", ou seja, vazia.
        self.topo = -1

    def esta_vazia(self):
        # Se o ponteiro ainda está em -1, não há nada na pilha.
        return self.topo == -1

    def esta_cheia(self):
        # Como o índice começa em 0, o último espaço válido é (capacidade - 1).
        return self.topo == self.capacidade - 1

    def push(self, elemento):
        # Proteção contra Stack Overflow.
        if self.esta_cheia():
            print("Erro: Stack Overflow! Pilha cheia.")
            return False

        # O segredo do Array Estático: nós não inserimos um novo item na lista,
        # nós apenas movemos o ponteiro uma casa para frente...
        self.topo += 1

        # ...e substituímos o valor 'None' (ou o valor antigo) pelo novo elemento
        # naquela posição específica do array.
        self.array[self.topo] = elemento
        print(f"Push: '{elemento}' adicionado ao topo. Topo em índice {self.topo}")
        return True

    def pop(self):
        # Proteção contra Stack Underflow.
        if self.esta_vazia():
            print("Erro: Stack Underflow! Pilha vazia.")
            return None

        # Pega o elemento que está na posição atual do ponteiro.
        elemento_removido = self.array[self.topo]

        # Limpar a posição (colocar None de volta) é opcional em Python, mas é uma
        # excelente prática! Evita que objetos fiquem presos na memória (Memory Leak)
        # se o array estiver segurando referências pesadas.
        self.array[self.topo] = None

        # Nós não deletamos o espaço do array, apenas movemos o ponteiro uma casa para trás.
        self.topo -= 1

        print(f"Pop: '{elemento_removido}' removido do topo. Novo topo em índice {self.topo}")
        return elemento_removido

    def ver_topo(self):
        # Permite "espiar" o elemento do topo sem movê-lo (função 'peek').
        if self.esta_vazia():
            return None
        return self.array[self.topo]

    def tamanho(self):
        # Como o índice começa em 0, o tamanho real é sempre o índice do topo + 1.
        return self.topo + 1

# ==========================================
# Testando a Pilha via Array Estático
# ==========================================
print("\n--- Teste da PilhaArrayEstatico ---")
# Criamos um array fixo: [None, None, None]
pilha = PilhaArrayEstatico(capacidade=3)

# O array vai mudando e o ponteiro vai subindo:
pilha.push(10) # Array: [10, None, None] | Topo: 0
pilha.push(20) # Array: [10, 20, None]   | Topo: 1
pilha.push(30) # Array: [10, 20, 30]     | Topo: 2

# O ponteiro chegou na (capacidade - 1). A pilha está cheia!
pilha.push(40) # Dispara o erro de Stack Overflow.

print("-" * 30)

print(f"Elemento no topo: {pilha.ver_topo()}") # Lê o índice 2 (30)
print(f"Tamanho da pilha: {pilha.tamanho()}")  # 2 + 1 = 3

print("-" * 30)

# O ponteiro vai descendo e limpando os rastros:
pilha.pop() # Remove 30. Array: [10, 20, None] | Topo: 1
pilha.pop() # Remove 20. Array: [10, None, None] | Topo: 0

print("-" * 30)

# Sobrescrevemos o próximo espaço livre:
pilha.push(50) # Array: [10, 50, None] | Topo: 1

print(f"Elemento no topo: {pilha.ver_topo()}") # Lê o índice 1 (50)
print(f"Tamanho da pilha: {pilha.tamanho()}")  # 1 + 1 = 2

print("-" * 30)

# Esvaziando até estourar para baixo:
pilha.pop() # Remove 50
pilha.pop() # Remove 10 (Topo volta para -1)
pilha.pop() # Topo é -1. Dispara o erro de Stack Underflow.
```

**02 | Fila Circular**

Missão: Otimize o uso de memória implementando uma fila sobre um array circular. Desenvolva enqueue() e dequeue(), controlando independentemente os ponteiros de início (front) e fim (rear).

```python
class FilaCircular:
    def __init__(self, capacidade):
        self.capacidade = capacidade
        self.array = [None] * capacidade

        # O ponteiro 'front' indica onde está o primeiro elemento a ser removido (o mais antigo).
        self.front = 0

        # O ponteiro 'rear' indica ONDE o PRÓXIMO elemento será inserido.
        self.rear = 0

        # Manter o controle do tamanho atual facilita muito a verificação
        # de fila cheia ou vazia sem precisarmos fazer matemática complexa com os ponteiros.
        self.tamanho = 0

    def esta_vazia(self):
        return self.tamanho == 0

    def esta_cheia(self):
        return self.tamanho == self.capacidade

    def enqueue(self, item):
        """Adiciona um item no final da fila (rear)."""
        if self.esta_cheia():
            print(f"Erro [Overflow]: Não é possível adicionar '{item}'. A fila está cheia.")
            return False

        # Insere o item na posição atual do ponteiro 'rear'
        self.array[self.rear] = item
        print(f"Enqueue: '{item}' inserido no índice {self.rear}.")

        # A MÁGICA DO ARRAY CIRCULAR ESTÁ AQUI:
        # Usamos o operador de módulo (%). Se a capacidade for 5 e o rear estiver em 4,
        # (4 + 1) % 5 = 0. O ponteiro "dá a volta" para o índice 0!
        self.rear = (self.rear + 1) % self.capacidade
        self.tamanho += 1

        return True

    def dequeue(self):
        """Remove e retorna o item do início da fila (front)."""
        if self.esta_vazia():
            print("Erro [Underflow]: A fila está vazia.")
            return None

        # Lê o elemento apontado pelo 'front'
        item_removido = self.array[self.front]

        # Limpa o espaço de memória (boa prática)
        self.array[self.front] = None
        print(f"Dequeue: '{item_removido}' removido do índice {self.front}.")

        # O ponteiro 'front' também dá a volta usando a mesma lógica do módulo
        self.front = (self.front + 1) % self.capacidade
        self.tamanho -= 1

        return item_removido

    def exibir_status(self):
        """Mostra a memória física e a posição dos ponteiros."""
        print(f"-> Memória: {self.array} | Front: {self.front} | Rear: {self.rear} | Tamanho: {self.tamanho}/{self.capacidade}\n")


# ==========================================
# Teste e Simulação do Array Circular
# ==========================================
print("--- Inicializando a Fila Circular (Capacidade: 3) ---")
fila = FilaCircular(capacidade=3)
fila.exibir_status()

print("--- Enchendo a Fila ---")
fila.enqueue("Chamado_A")
fila.enqueue("Chamado_B")
fila.enqueue("Chamado_C")
fila.exibir_status()

# Tentativa de exceder o limite físico
fila.enqueue("Chamado_D")

print("\n--- Processando e Liberando Espaço ---")
fila.dequeue() # Remove Chamado_A (Índice 0 fica livre)
fila.exibir_status()

print("--- O Ponteiro 'Rear' dá a volta ---")
# Como o índice 0 foi liberado, o Chamado_D vai ocupar esse espaço no início do array,
# mesmo entrando por último!
fila.enqueue("Chamado_D")
fila.exibir_status()

print("--- Processando o Restante ---")
fila.dequeue() # Remove Chamado_B do índice 1
fila.dequeue() # Remove Chamado_C do índice 2
fila.exibir_status()

print("--- O Ponteiro 'Front' também dá a volta ---")
fila.dequeue() # Remove Chamado_D do índice 0 (O front volta pro início)
fila.exibir_status()
```
